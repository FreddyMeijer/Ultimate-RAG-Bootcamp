## Introductie
Voordat we een document kunnen opnemen in een vector database, moeten we het document klaar maken. Dit doen we via de LangChain document structure. We verrijken de data met metadata alvorens de data om te zetten naar vectoren. Dus je krijgt iets zoals:

```python
# creating a document

try:
    doc = Document(
        page_content="This is a sample document created using the Document class.",
        metadata={"source": "chapter1.pdf", 
                "page": "5",
                "timestamp": "2023-10-01T12:00:00Z"
                }
        
    )
except NameError:
    print(" ")
```

In bovenstaand voorbeeld hebben we een errorhandler opgenomen om errors niet weer te geven. Bovenstaande code genereert een `NameError` omdat we niet de juiste pakketten hebben geimporteerd. 

We zien twee belangrijke elementen terug: `page_content` en `metadata`. De page_content is de daadwerkelijke inhoud van het document. Het bevat de informatie die opgenomen moeten worden in de vectordatabase en die doorzoekbaar moet zijn. De inhoud van page_content wordt altijd opgeslagen als datatype string. Het is van belang om voor het opslaan alle onnodige formatering weg te halen en na te denken over de chunksize (normaal gesproken 500-2000 tokens).

De metadate geeft aanvullende informatie over het document. De metadata wordt later gebruikt voor het filteren, volgen en bepalen van de context van het document. Het bevat HSON data. Veel voorkomende metadata velden zijn:

|Veld|Gebruik|Voorbeeld|
| --- | --- | --- | 
|source|bestandspad of URL|`"docs/manuals.pdf"`|
|page / chunk_id|locatie in het bestand|`page: 42, chunk: 7`|
|timestamp|aanmaakdatum|`"2023-10-01T12:00:00Z"`|
|author|auteur|`"John Doe"`|
|catagory / type|classificatie|`"technical", "legal"`|
|language|content taal|`"en", "es", "fr"`|

Het staat je vrij om 'eigen' velden toe te voegen zoals bijvoorbeeld `departement`, `security_level` en `version`.
