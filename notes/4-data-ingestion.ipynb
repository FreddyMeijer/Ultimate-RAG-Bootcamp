# Module overzicht

In dit kennisitem wordt alles beschreven dat te maken heeft met de `parsing` en `ingestion` van data voor RAG systemen. Dit doen we met normale tekstbestanden en met complexere PDF's en databases. Er wordt gebruik gemaakt van LangChain. In dit kennisitem zien we verschillende technieken, ondersteund met praktische voorbeelden.

## Inhoudsopgave
1. Introductie
2. Tekstbestanden (.txt)
3. Markdown bestanden (.md)
4. PDF bestanden (.pdf)
5. Microsoft Word (.docx)
6. CSV en Excel documenten (.csv/.xlsx)
7. JSON en andere gestructureerde bestanden (.json/.xml)
8. Webscraping
9. Databases (SQL)
10. Audio en video transcripties
11. Geavanceerde technieken
12. Best practices
